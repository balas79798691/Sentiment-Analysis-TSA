# Review Sentiment Analyzer

A text-analysis application that classifies reviews as positive, negative, or neutral.

## 1. Libraries

```python
import re
import string
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob

print("Review Sentiment Analyzer is ready.")
```

## 2. Text Preprocessing

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"@\w+|#\w+", "", text)
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = re.sub(r"\d+", "", text)
    return re.sub(r"\s+", " ", text).strip()

sample = "The product is AMAZING! I really loved its quality and performance."
print("Original:", sample)
print("Cleaned :", clean_text(sample))
```

## 3. Sentiment Engine

```python
def analyze_sentiment(text):
    analysis = TextBlob(text)
    polarity = analysis.sentiment.polarity
    subjectivity = analysis.sentiment.subjectivity

    if polarity > 0.1:
        label = "Positive"
    elif polarity < -0.1:
        label = "Negative"
    else:
        label = "Neutral"

    return label, round(polarity, 3), round(subjectivity, 3)

print(analyze_sentiment(sample))
```

## 4. Application

```python
def sentiment_analyzer():
    print("=" * 55)
    print("          REVIEW SENTIMENT ANALYZER")
    print("=" * 55)
    print("Type 'quit' to exit.\n")

    while True:
        text = input("Enter a review: ")

        if text.lower() == "quit":
            print("Application closed.")
            break

        if not text.strip():
            print("Please enter some text.\n")
            continue

        sentiment, polarity, subjectivity = analyze_sentiment(text)

        print("\n--- RESULT ---")
        print("Sentiment    :", sentiment)
        print("Polarity     :", polarity)
        print("Subjectivity :", subjectivity)

        if sentiment == "Positive":
            print("Interpretation: favorable opinion")
        elif sentiment == "Negative":
            print("Interpretation: unfavorable opinion")
        else:
            print("Interpretation: neutral opinion")

        print("-" * 55)

# Run this cell to use the application:
# sentiment_analyzer()
```

## 5. Test Cases

```python
reviews = [
    "The movie was fantastic and enjoyable.",
    "The service was terrible and rude.",
    "The phone has 128 GB of storage.",
    "I am very happy with this purchase.",
    "The product stopped working after two days."
]

records = []
for review in reviews:
    s, p, sub = analyze_sentiment(review)
    records.append({"Review": review, "Sentiment": s, "Polarity": p, "Subjectivity": sub})

df = pd.DataFrame(records)
display(df)

plt.figure(figsize=(7,4))
df["Sentiment"].value_counts().plot(kind="bar")
plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Conclusion

This application cleans review text, calculates polarity and subjectivity, classifies sentiment, and visualizes the results.
